```python
df = pd.read_csv('final_full_merged_dataset.csv')

mental_health_change = df.groupby('GEOID')['Mental Health'].agg(lambda x: x.iloc[-1] - x.iloc[0])
top_geo_id = mental_health_change.idxmax()

target_df = df[df['GEOID'] == top_geo_id]

factors = [
    'Binge Drinking',
    '%with disability 18-64',
    '%with disability 65+',
    '%cognitive disability 18-64',
    '%cognitive disability 65+',
    '%self-care disability 18-64',
    '%self-care disability 65+',
    '%independent living difficulty 18-64',
    '%independent living difficulty 65+',
    'now married ',
    'widowed',
    'divorced',
    'seperated',
    'never married ',
    'high school degree or higher ',
    'bachelors degree or higher',
    'families and people below poverty level',
    'Civilian labor force employed',
    'Civilian labor force unemployed',
    'Population'
]

melted_df = target_df.melt(id_vars='Year', value_vars=factors, var_name='Factor', value_name='Value')

melted_df['Value'] = (
    melted_df['Value']
    .astype(str)
    .str.replace(',', '', regex=False)
    .str.replace('%', '', regex=False)
)

melted_df['Value'] = pd.to_numeric(melted_df['Value'], errors='coerce')

melted_df = melted_df.dropna(subset=['Value'])

variation = melted_df.groupby('Factor')['Value'].nunique()
dynamic_factors = variation[variation > 1].index.tolist()
filtered_df = melted_df[melted_df['Factor'].isin(dynamic_factors)]

g = sns.FacetGrid(filtered_df, col='Factor', col_wrap=4, sharex=True, sharey=False, height=3.5)

g.map_dataframe(sns.lineplot, x='Year', y='Value', marker='o')

for ax in g.axes.flatten():
    lines = ax.get_lines()
    if lines:
        data = lines[0].get_xydata()
        for x, y in data:
            ax.text(x, y, f"{y:.1f}", ha='center', va='bottom', fontsize=7)

plt.figtext(0.5, -0.02, "Note: Factors with no variation over time have been removed from this visualisation.", wrap=True, horizontalalignment='center', fontsize=9)

plt.tight_layout()
plt.show()
```

***Testing the Alignement of Poverty and Mental Health Among States***

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('final_full_merged_dataset.csv')

mh_col = 'Mental Health'
poverty_col = 'families and people below poverty level'

df_filtered = df[['StateDesc', 'Year', mh_col, poverty_col]]
df_filtered = df_filtered[df_filtered['Year'].isin([2014, 2022])]
df_filtered[mh_col] = pd.to_numeric(df_filtered[mh_col], errors='coerce')
df_filtered[poverty_col] = pd.to_numeric(df_filtered[poverty_col], errors='coerce')
df_filtered = df_filtered.dropna()

state_corrections = {
    'District of C': 'District of Columbia',
    'North Carolin': 'North Carolina',
    'South Carolin': 'South Carolina',
    'District of Columbia': 'District of Columbia'
}
df_filtered['StateDesc'] = df_filtered['StateDesc'].replace(state_corrections)

grouped = df_filtered.groupby(['StateDesc', 'Year']).agg({
    mh_col: 'mean',
    poverty_col: 'mean'
}).reset_index()

pivot_df = grouped.pivot(index='StateDesc', columns='Year')
pivot_df.columns = ['MH_2014', 'MH_2022', 'Pov_2014', 'Pov_2022']
pivot_df = pivot_df.reset_index()

pivot_df['Poverty Change (%)'] = ((pivot_df['Pov_2022'] - pivot_df['Pov_2014']) / pivot_df['Pov_2014']) * 100

threshold = 10

def classify(row):
    if row['Poverty Change (%)'] <= -threshold:
        return 'Low Poverty (Significant Decrease)'
    else:
        return 'High Poverty (No Significant Decrease)'

pivot_df['Poverty Category'] = pivot_df.apply(classify, axis=1)

pivot_df['MH Change'] = pivot_df['MH_2022'] - pivot_df['MH_2014']
pivot_df['MH Category'] = pivot_df['MH Change'].apply(lambda x: 'Improved' if x > 0 else 'Worsened')

def structural_alignment(row):
    if row['Poverty Category'] == 'Low Poverty (Significant Decrease)' and row['MH Category'] == 'Improved':
        return 'Low Poverty, High MH (misaligned)'
    elif row['Poverty Category'] == 'Low Poverty (Significant Decrease)' and row['MH Category'] == 'Worsened':
        return 'Low Poverty, Low MH (dual decline)'
    elif row['Poverty Category'] == 'High Poverty (No Significant Decrease)' and row['MH Category'] == 'Improved':
        return 'High Poverty, High MH (aligned)'
    else:
        return 'High Poverty, Low MH (unexpected gain)'

pivot_df['Structural Category'] = pivot_df.apply(structural_alignment, axis=1)

category_counts = pivot_df['Structural Category'].value_counts().sort_values()

plt.figure(figsize=(10, 6))
bars = plt.barh(category_counts.index, category_counts.values, color='steelblue')
plt.xlabel('Number of States')
plt.title('Model 7: Structural Alignment Categories (2014–2022)')
plt.grid(axis='x', linestyle='--', alpha=0.5)

for bar in bars:
    width = bar.get_width()
    plt.text(width + 0.5, bar.get_y() + bar.get_height() / 2,
             f'{int(width)} states', va='center', fontsize=10)

plt.tight_layout()
plt.show()

print("States in Each Structural Category:\n")
for cat in category_counts.index:
    states = pivot_df[pivot_df['Structural Category'] == cat]['StateDesc'].tolist()
    print(f"{cat} ({len(states)} states):")
    print(", ".join(sorted(states)))
    print()
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.linear_model import Ridge, Lasso
from sklearn.neural_network import MLPRegressor
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, mean_absolute_percentage_error

df = pd.read_csv('final_full_merged_dataset.csv', low_memory=False)

df = df.copy()
for col in df.columns:
    if df[col].dtype == object:
        df[col] = df[col].str.replace(',', '', regex=False).str.replace('%', '', regex=False)

df = df.apply(pd.to_numeric, errors='coerce')

df = df.fillna(df.median(numeric_only=True))

drop_cols = ['GEOID', 'CountyName', 'StateDesc', 'Year', 'Data_Value_Type']
df = df.drop(columns=drop_cols, errors='ignore')

target = 'Mental Health'
X = df.drop(columns=[target], errors='ignore')
y = df[target]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

models = {
    "Random Forest": RandomForestRegressor(n_estimators=100, random_state=42),
    "XGBoost": XGBRegressor(n_estimators=100, random_state=42, verbosity=0),
    "Gradient Boosting": GradientBoostingRegressor(
        n_estimators=100, learning_rate=0.1, max_depth=3,
        min_samples_leaf=5, subsample=0.8, random_state=42
    ),
    "MLP Regressor": MLPRegressor(hidden_layer_sizes=(50,), max_iter=300, random_state=42),
    "Ridge Regression": Ridge(alpha=1.0),
    "Lasso Regression": Lasso(alpha=0.01),
}

results = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    preds = model.predict(X_test)
    results[name] = {
        "R² Score": round(r2_score(y_test, preds), 4),
        "MAE": round(mean_absolute_error(y_test, preds), 4),
        "MSE": round(mean_squared_error(y_test, preds), 4),
        "MAPE": round(mean_absolute_percentage_error(y_test, preds), 4)
    }


results_df = pd.DataFrame(results).T
print(" Final Model Evaluation:\n")
print(results_df)
```
